# Flash Flood Prediction System - Exploratory Data Analysis (EDA)
## Hilly Regions Multi-Source Telemetry & Hydrological Inundation Study

This notebook analyzes multi-source data including:
- Digital Elevation Models (DEM) & Slope Steepness
- Satellite Radar Precipitation Intensities & 24h Accumulation
- IoT Multi-Layer Soil Moisture Saturation Levels
- Acoustic Doppler River Hydrometry (Water Stage & Velocity)
- Catchment Drainage Morphology & NDVI Buffering

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.utils import generate_multi_source_dataset

sns.set_theme(style='darkgrid')
print('Environment initialized.')
```

```python
# Load or generate dataset
df = generate_multi_source_dataset(n_samples=3000, random_state=42)
print('Dataset Dimensions:', df.shape)
df.head()
```

```python
# Statistical Overview
df.describe().T
```

```python
# Correlation Matrix
plt.figure(figsize=(12, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Multi-Source Hydrological Feature Correlations')
plt.show()
```

```python
# Rainfall vs River Level vs Flood Severity Distribution
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.histplot(df['rainfall_intensity_mm_hr'], kde=True, ax=axes[0], color='#06b6d4')
axes[0].set_title('Rainfall Intensity (mm/hr)')

sns.histplot(df['soil_moisture_percentage'], kde=True, ax=axes[1], color='#f59e0b')
axes[1].set_title('Soil Moisture (%)')

sns.histplot(df['flood_risk_severity_index'], kde=True, ax=axes[2], color='#ef4444')
axes[2].set_title('Flash Flood Severity Index')
plt.show()
```
